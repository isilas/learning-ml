## 1. Simulation design
Draw 1,000 independent points uniformly in $[0,1]^d$ for each dimension. Exclude self-distances. Compare mean nearest-neighbor distance to the median pairwise distance, which gives a scale-relative measure.

```python
from projects.fundamentals import dimension_experiment
rows=dimension_experiment(seed=0,n=1000)
for row in rows: print(row)
```

```python
d=[r['dimension'] for r in rows]
fig,axes=plt.subplots(1,2,figsize=(11,4))
axes[0].errorbar(d,[r['mean'] for r in rows],yerr=[r['std'] for r in rows],marker='o',capsize=3)
axes[0].set(xscale='log',xlabel='Dimensions',ylabel='Nearest-neighbor distance')
axes[1].plot(d,[r['relative_to_median_pair_distance'] for r in rows],marker='o')
axes[1].set(xscale='log',xlabel='Dimensions',ylabel='Nearest / typical distance',ylim=(0,1))
fig.tight_layout(); plt.show()
```

## 2. Interpretation
The recorded ratio rises from 0.002 in 1D to 0.818 in 100D. Nearest points become less distinct relative to a typical pairwise distance. This motivates feature selection, dimensionality reduction and domain-informed metrics. Results apply to this synthetic distribution and fixed sample size.
